```python
# 1. Import library yang diperlukan
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Load dataset
df = pd.read_excel('../data/raw/dataset_1000.xlsx')

# 3. Lihat struktur data
df.head()
df.info()
df.describe()

# 4. Cek missing values
df.isna().sum()

# 5. Cek distribusi Waste Flag (target ML)
sns.countplot(x='Waste_Flag', data=df)

# 6. Analisis Korelasi
corr = df.corr(numeric_only=True)  # hanya kolom numerik
plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()

# 7. Analisis fitur penting (quick insight)
sns.boxplot(x='Waste_Flag', y='Expiry_Days_Left', data=df)
```
